# 📑 Sub-Materi 1: Variabel, Tipe Data, Input, & Output
**Tujuan:** Memahami penggunaan variabel, konversi tipe data, serta interaksi input/output dasar.

---

### 💡 Penjelasan Ringkas:
* **Variabel**: Wadah penyimpanan data di memori.
* **Tipe Data**:
  * `str` (String): Teks, diapit tanda petik.
  * `int` (Integer): Angka bulat.
  * `float` (Float): Angka desimal.
  * `bool` (Boolean): Nilai logika (`True` atau `False`).
* **Input & Output**: `input()` menerima masukan teks dari pengguna, sedangkan `print()` menampilkan hasil ke layar.
* **Type Casting**: Mengubah tipe data, contohnya `int("10")` mengubah teks `"10"` menjadi angka 10.

---

### 📝 Catatan Rangkuman Syntax:
- `input()` -> Menerima masukan dari pengguna (hasilnya selalu bertipe teks/string).
- `print()` -> Menampilkan data atau teks ke layar console.
- `int()` / `float()` -> Melakukan type casting (mengubah teks angka menjadi tipe numeric).
- `f"teks {variabel}"` -> Format f-string untuk menyisipkan variabel langsung ke dalam teks secara rapi.


```python
# 1. Menerima input dari pengguna
nama_lengkap = input("Masukkan nama lengkap: ")
umur = int(input("Masukkan umur kamu saat ini: "))
ipk_target = float(input("Masukkan target IPK semester 1: "))

# 2. Pemrosesan data (Hitung estimasi tahun lahir)
tahun_lahir= 2026 - umur

# 3. Menampilkan Output (menggunakan f-string)
print("\n=== DATA MAHASISWA ===")
print(f"Nama Lengkap: {nama_lengkap}")
print(f"Tahun Lahir: {tahun_lahir}")
print(f"IPK Target: {ipk_target}")
```

# 📑 Sub-Materi 2: Operator & Logika Percabangan (`if-elif-else`)
**Tujuan:** Memahami logika pengambilan keputusan dalam program berdasarkan kondisi tertentu.

---

### 💡 Penjelasan Ringkas:
* **Operator Perbandingan**: Digunakan untuk membandingkan nilai (`>`, `<`, `>=`, `<=`, `==` untuk sama dengan, `!=` untuk tidak sama dengan).
* **Percabangan (`if-elif-else`)**:
  * `if`: Kondisi utama yang pertama kali diperiksa.
  * `elif` *(else if)*: Kondisi alternatif jika kondisi sebelumnya salah (`False`).
  * `else`: Kondisi terakhir jika semua kondisi di atasnya salah.
* **Indentasi (Spasi)**: Python menggunakan spasi/tab di awal baris untuk menandai blok kode yang berada di dalam perintah `if/elif/else`.

---

### 📝 Catatan Rangkuman Syntax:
- `>` / `<` / `==` -> Operator pembanding untuk mengevaluasi nilai (menghasilkan `True` atau `False`).
- `if kondisi:` -> Memulai blok keputusan.
- `elif kondisi:` -> Menambahkan cabang kondisi alternatif.
- `else:` -> Menentukan aksi default jika tidak ada kondisi yang terpenuhi.

```python
# 1. Mengambil input angka nilai dari pengguna
nilai_matkul = float(input("Masukkan nilai akhir Algoritma & Pemrograman (0-100): ").replace(',', '.'))

# 2. Logika Percabangan untuk Menentukan Indeks Huruf & Predikat
if nilai_matkul >= 80:
  indeks = "A"
  status = "Lulus Sangat Memuaskan (Cumlaude)"
elif nilai_matkul >= 70:
  indeks = "B"
  status = "Lulus Memuaskan"
elif nilai_matkul >= 60:
  indeks = "C"
  status = "Lulus Cukup"
elif nilai_matkul >= 50:
  indeks = "D"
  status = "Lulus Bersyarat"
else:
  indeks = "E"
  status = "Tidak Lulus (Wajib Mengulang)"

# 3. Menampilkan Hasil Evaluasi
print("\n=== HASIL EVALUASI MATKUL ===")
print(f"Nilai Akhir: {nilai_matkul}")
print(f"Indeks Huruf: {indeks}")
print(f"Predikat: {status}")
```

# 📑 Sub-Materi 3: Perulangan (`for` & `while`)
**Tujuan:** Memahami penggunaan perulangan untuk mengeksekusi blok kode secara berulang-ulang.

---

### 💡 Penjelasan Ringkas:
* **`for` Loop**: Digunakan ketika kita sudah tahu berapa kali perulangan akan dilakukan (misalnya mengulang sebanyak $N$ kali menggunakan `range()`).
* **`while` Loop**: Digunakan ketika perulangan terus berjalan selama suatu kondisi masih bernilai `True` (berhenti ketika kondisi menjadi `False`).
* **`range(start, stop, step)`**: Fungsi pembantu `for` loop untuk menghasilkan urutan angka.

---

### 📝 Catatan Rangkuman Syntax:
- `for i in range(n):` -> Mengulang blok kode sebanyak `n` kali (dimulai dari 0 sampai `n-1`).
- `while kondisi:` -> Mengulang blok kode terus-menerus selama kondisi bernilai `True`.
- `break` -> Perintah untuk menghentikan dan keluar dari perulangan secara paksa.

```python
# 1. Contoh for loop: Menampilkan daftar praktikum semester 1

print("=== JADWAL PRAKTIKUM MINGGUAN")
for minggu in range (1, 6): # loop dari angka 1 sampai 5
    print(f"Minggu ke-{minggu}: Latihan Algoritma & Pemrograman")

print("\n" + "="*35 + "\n")

# 2. Contoh while loop: Simulasi percobaan login sederhana
password_benar = "telkom2026"
percobaan = 0
maks_percobaan = 3

while percobaan < maks_percobaan:
  input_user = input(f"Masukan Password (Percobaan {percobaan + 1}/{maks_percobaan}): ")

  if input_user == password_benar:
    print("✅ Login Berhasil! Selamat datang di Portal Mahasiswa.")
    break # Keluar dari loop karena password benar
  else:
      print("❌ Password salah, coba lagi.")
      percobaan += 1

if percobaan == maks_percobaan:
    print("\n⚠️ Akun terblokir sementara karena salah password 3 kali.")
```

# 📑 Sub-Materi 4: Function (Fungsi) & Variabel Scope
**Tujuan:** Memahami pembuatan blok kode yang reusable menggunakan fungsi, parameter, dan nilai kembalian (return value).

---

### 💡 Penjelasan Ringkas:
* **Function (`def`)**: Blok kode terorganisir yang dibuat untuk menjalankan satu tugas tertentu.
* **Parameter / Argument**: Nilai yang dimasukkan ke dalam fungsi sebagai bahan olahan.
* **`return`**: Perintah untuk mengembalikan hasil proses fungsi ke pemanggilnya.
* **Scope**: Variabel di dalam fungsi (lokal) tidak bisa diakses langsung dari luar fungsi.

---

### 📝 Catatan Rangkuman Syntax:
- `def nama_fungsi(parameter):` -> Merekam/mendefinisikan fungsi baru.
- `return hasil` -> Mengirimkan nilai hasil kalkulasi keluar dari fungsi.
- `nama_fungsi(argumen)` -> Memanggil dan menjalankan fungsi yang telah dibuat.

```python
# 1. Mendefinisikan fungsi untuk menghitung Nilai Akhir Mata Kuliah
def hitung_nilai_akhir(tugas, uts, uas):
    """Fungsi ini menghitung bobot nilai: 30% Tugas + 30% UTS + 40% UAS"""
    nilai_total = (tugas * 0.3) + (uts * 0.3) + (uas * 0.4)
    return nilai_total

# 2. Mendefinisikan fungsi untuk mengecek status kelulusan
def cek_status(nilai):
    if nilai >= 60:
        return "LULUS"
    else:
        return "TIDAK LULUS"

# 3. Memanggil dan menggunakan fungsi
nil_tugas = float(input("Masukan Nilai Tugas (0-100): ").replace(',', '.'))
nil_uts = float(input("Masukan Nilai UTS (0-100): ").replace(',', '.'))
nil_uas = float(input("Masukan Nilai UAS (0-100): ").replace(',', '.'))

# Eksekusi fungs
total_nilai = hitung_nilai_akhir(nil_tugas, nil_uts, nil_uas)
status_akhir = cek_status(total_nilai)

# Menampilkan Hasil
print("\n=== REKAPITULASI NILAI FUNGSI ===")
print(f"Nilai Akhir Kalkulasi: {total_nilai:.2f}")
print(f"Status Kelulusan     : {status_akhir}")
```

# 📑 Sub-Materi 5: Array / List & Manipulasi Data Sederhana
**Tujuan:** Memahami cara menyimpan, mengakses, dan mengolah kumpulan data (koleksi) dalam satu variabel menggunakan List/Array.

---

### 💡 Penjelasan Ringkas:
* **List (Array)**: Tipe data terurut yang digunakan untuk menyimpan kumpulan nilai/elemen. Ditulis menggunakan kurung siku `[]`.
* **Index**: Posisi elemen di dalam List, **selalu dimulai dari angka 0** (elemen pertama = index 0).
* **Fungsi Agregasi**: Fungsi bawaan Python untuk mengolah List angka, seperti `sum()` (jumlah), `len()` (banyaknya elemen), `max()` (nilai tertinggi), dan `min()` (nilai terendah).

---

### 📝 Catatan Rangkuman Syntax:
- `nama_list = [val1, val2, val3]` -> Membuat List baru.
- `nama_list[0]` -> Mengakses elemen pertama dalam List.
- `nama_list.append(nilai)` -> Menambahkan elemen baru ke urutan paling akhir List.
- `len(nama_list)` -> Menghitung total banyaknya elemen dalam List.
- `sum(nama_list) / len(nama_list)` -> Rumus mencari rata-rata elemen angka.

```python
# 1. Membuat List berisi daftar nilai tugas mahasiswa
daftar_nilai = [80, 90, 65, 50, 95]

# 2. Menampilkan data awal dan mengakses elemen tertentu
print("=== PENGOLAHAN DATA NILAI (LIST) ===")
print(f"Daftar Nilai Awal: {daftar_nilai}")
print(f"Nilai Tugas Pertama (Index 0): {daftar_nilai[0]}")
print(f"Nilai Tugas Terakhir         : {daftar_nilai[-1]}")

# 3. Menambahkan nilai baru menggunakan .append()
nilai_baru = float(input("\nMasukkan Nilai Susulan (0-100): ").replace(',', '.'))
daftar_nilai.append(nilai_baru)

# 4. Kalkulasi Agregasi Data
total_nilai = sum(daftar_nilai)
banyak_tugas = len(daftar_nilai)
rata_rata = total_nilai / banyak_tugas
nilai_tertinggi = max(daftar_nilai)
nilai_terendah = min(daftar_nilai)

# 5. Menampilkan Hasil Olahan
print("\n=== HASIL STATISTIK NILAI ===")
print(f"Daftar Nilai Terbaru : {daftar_nilai}")
print(f"Total Jumlah Nilai   : {total_nilai}")
print(f"Banyaknya Tugas      : {banyak_tugas}")
print(f"Nilai Rata-Rata      : {rata_rata:.2f}")
print(f"Nilai Tertinggi      : {nilai_tertinggi}")
print(f"Nilai Terendah       : {nilai_terendah}")
```
